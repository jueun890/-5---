01. 이진 검색 트리의 검색 시간
이진 검색 트리에서 균형이 맞지 않는 경우, 트리의 형태에 따라 검색 성능은 다음과 같습니다.

최악의 경우: 트리가 한쪽으로 치우친 편향 트리일 때, $O(n)$의 시간이 걸립니다.

최선의 경우: 트리가 완전한 균형을 이룰 때, $O(\log n)$의 시간이 걸립니다.

04. 
03번 트리는 LR 불균형 상태이며, 이를 해결하기 위해 이중 회전을 수행합니다.

수선 과정
1.Left Rotation (노드 10-20 구간): 노드 20을 노드 10의 부모 위치로 올립니다. (LL 상태로 변환)

2.Right Rotation (노드 40 기준): 노드 20을 루트로 올리고 노드 40을 오른쪽 자식으로 내립니다.

```python
class Node:
    def __init__(self, key):
        self.key = key
        self.left = None
        self.right = None

root = Node(20)

root.left = Node(10)
root.left.left = Node(5)
root.left.right = Node(17)

root.right = Node(40)
root.right.right = Node(55)

def print_tree(node, level=0):
    if node:
        print_tree(node.right, level + 1)
        print(' ' * 8 * level + '->', node.key)
        print_tree(node.left, level + 1)

print_tree(root)
```

05. 
LR 케이스, RL 케이스

이유: LL과 RR은 한 번의 회전으로 끝나지만, LR은 Left-Right 회전, RL은 Right-Left 회전으로 총 두 번의 회전이 필요하다.

06. 틀린 주장

이유: 수선 후 트리의 균형이 잡히면서 높이가 낮아지는 것이 일반적이지만, 삽입/삭제가 일어난 직후의 트리 높이가 수선 전 원래의 정상적인 AVL 트리 높이와 같아지는 경우도 많습니다. 즉, 반드시 낮아지는 것이 아니라 불균형이 해소되어 유지되거나 낮아지는 것이다.

07. 발생 가능성 있다

삽입 시: 삽입 시에는 한 번의 수선(단일 또는 이중 회전)으로 전체 트리의 균형이 완료됩니다.

삭제 시: 삭제의 경우, 특정 노드를 삭제하고 회전하여 균형을 맞췄더라도 그 상위 노드에서 다시 불균형이 발생할 수 있습니다. 이 과정이 루트까지 반복될 수 있으므로, 트리의 높이에 따라 3회 이상의 수선이 충분히 발생할 수 있습니다.

```python
class RBNode:
    def __init__(self, key, color="Red"):
        self.key = key
        self.color = color  # "Red" or "Black"
        self.left = None
        self.right = None

def print_rb_tree(node, level=0, prefix="Root: "):
    """RB-트리의 구조와 색상을 시각적으로 출력합니다."""
    if node:
        print("  " * level + f"{prefix}{node.key} ({node.color})")
        if node.left or node.right:
            if node.left:
                print_rb_tree(node.left, level + 1, "L-- ")
            else:
                print("  " * (level + 1) + "L-- None (Black)")
            if node.right:
                print_rb_tree(node.right, level + 1, "R-- ")
            else:
                print("  " * (level + 1) + "R-- None (Black)")


root = RBNode(10, "Black")


node_5 = RBNode(5, "Red")
root.left = node_5

node_3 = RBNode(3, "Black")
node_3.left = RBNode(2, "Red")
node_5.left = node_3

node_6 = RBNode(6, "Black") # Recolored to Black
node_8 = RBNode(8, "Black") # Recolored to Black
node_8.left = RBNode(7, "Red") # 신규 삽입된 키 7
node_6.right = node_8
node_5.right = node_6

node_13 = RBNode(13, "Black")
node_13.right = RBNode(15, "Red")
root.right = node_13

print("--- 8번 문제: 키 7 삽입 및 Recoloring 완료 후 RB-트리 ---")
print_rb_tree(root)
```

09. RB 트리 최악의 상황이 깨졌을 때, 수선 과정이 루트까지 올라가는 상황은 Recoloring이 반복될 때입니다. 삽입된 노드 $p$의 부모와 삼촌이 모두 Red일 때 Recoloring을 하면, 조부모 노드가 Red가 됩니다. 이때 조부모의 부모(증조부모)가 다시 Red라면 또다시 Double Red가 발생합니다.$p^6$까지의 과정: 이처럼 삼촌 노드가 계속 Red인 경로를 따라 올라가면, $p$에서 시작된 Red 전이가 $p^2, p^4, p^6$ 등 조상 노드로 계속 전파됩니다. 즉, $p$에서 시작된 불균형이 Recoloring만으로 해결되지 않고 계속 위로 전파되어 루트($p^k$)까지 도달하면 루트를 Black으로 바꾸며 전체적인 블랙 높이가 1 증가하고 종료됩니다.

10번 트리

```python
class BTreeNode:
    def __init__(self, keys=None, leaf=False):
        self.keys = keys if keys else []
        self.child = []
        self.leaf = leaf

def display_tree(node, indent=""):
    """B-트리의 구조를 계층적으로 출력합니다."""
    print(indent + str(node.keys))
    if not node.leaf:
        for c in node.child:
            display_tree(c, indent + "    ")


root = BTreeNode([25], leaf=False)


level1_left = BTreeNode([6, 13, 19], leaf=False)
level1_right = BTreeNode([34, 40], leaf=False)
root.child = [level1_left, level1_right]


leaf_1 = BTreeNode([1, 2, 3, 4, 5], leaf=True)
leaf_2 = BTreeNode([7, 8, 9, 10], leaf=True)
leaf_3 = BTreeNode([15, 17, 18], leaf=True) 
leaf_4 = BTreeNode([20, 23], leaf=True)     

level1_left.child = [leaf_1, leaf_2, leaf_3, leaf_4]

leaf_5 = BTreeNode([27, 28, 30, 31, 33], leaf=True)
leaf_6 = BTreeNode([35, 36, 37, 38, 39], leaf=True)
leaf_7 = BTreeNode([41, 45], leaf=True)

level1_right.child = [leaf_5, leaf_6, leaf_7]

print("--- 10번 문제: 키 18 삽입 후 최종 B-트리 구조 ---")
display_tree(root)
```

14. 내부 노드 원소의 직후 원소가 항상 리프에 있는 이유B-트리에서 특정 원소 $x$의 직후 원소(In-order Successor)는 $x$의 오른쪽 포인터를 타고 내려가서 만나는 서브트리의 가장 왼쪽 리프 노드의 첫 번째 원소이기 때문입니다. B-트리는 모든 리프 노드가 동일한 레벨에 존재하는 완전 균형 트리 구조를 유지하므로, 탐색 경로의 끝은 항상 리프 노드에 도달하게 됩니다.

16.
동일한 제약 조건: B-트리의 정의에 따라 리프 노드와 내부 노드(루트 제외) 모두 최소 $\lceil k/2 \rceil$개에서 최대 $k$개의 키를 가질 수 있다는 동일한 수적 범위를 공유합니다.자식 노드와의 연결성: 내부 노드는 $n$개의 키를 가질 때 항상 $n+1$개의 자식 노드를 가리키는 포인터를 가집니다. 반면, 리프 노드는 더 이상 자식이 없으므로 키만 저장하며 트리의 가장 하단 레벨에 위치합니다.구조적 균형: 모든 리프 노드는 항상 같은 레벨(깊이)에 존재하며, 내부 노드에 있는 키의 배치는 리프 노드로 가는 탐색 경로를 결정하는 기준점이 됩니다.삽입과 분할의 연쇄성: 리프 노드에 키가 가득 차서($k$개 초과) 분할(Split)이 일어나면, 그 중앙값(Median)이 상위 내부 노드로 올라가 내부 노드의 키 수를 하나 증가시킵니다. 이 과정이 반복되면 내부 노드 역시 분할되어 트리의 높이가 높아질 수 있습니다.

17.
$k$가 짝수인 경우: $2 \times (k/2) - 2 = k - 2$개가 되어 최대 용량 $k$보다 작습니다.$k$가 홀수인 경우: $2 \times ((k+1)/2) - 2 = k - 1$개가 되어 최대 용량 $k$보다 작습니다.따라서 병합된 노드의 키 개수는 항상 최대 용량 $k$를 넘지 않으므로, 병합으로 인해 오버플로우가 발생하는 상황은 구조적으로 불가능합니다.
